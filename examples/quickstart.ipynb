# PyUpSet: быстрый старт

Установите пакет в окружение ядра Jupyter: `%pip install -e ".[notebook]"` из корня проекта.
Каждая строка — наблюдение, каждый бинарный столбец — принадлежность к множеству.

```python
import pandas as pd
import matplotlib.pyplot as plt
import pyupset

df = pd.DataFrame({
    "A": [1, 1, 0, 1, 0, 1],
    "B": [0, 1, 1, 1, 0, 1],
    "C": [0, 1, 0, 1, 1, 0],
    "D": [0, 0, 1, 1, 0, 0],
})
df
```

## Сортировка по степени

Сверху слева — размеры множеств, снизу слева — матрица, справа — мощности точных пересечений. Цвет обозначает степень, серые точки — отсутствие признака.

```python
fig, axes = pyupset.plot(df, sets=["A", "B", "C", "D"], sort_by="degree")
fig
```

## Сортировка по мощности и настройка matplotlib

```python
styled, styled_axes = pyupset.plot(
    df, sets=list(df.columns), sort_by="size", figsize=(11, 7),
    colors={"intersections": "steelblue", "sets": "seagreen"},
)
styled_axes["intersections"].set_title("Точные пересечения")
styled
```

## Сохранение

Сохранение выполняет вызывающий код; сама библиотека не записывает файлы.

```python
styled.savefig("upset.png", dpi=300, bbox_inches="tight")
styled.savefig("upset.pdf", bbox_inches="tight")
```

## Автоопределение столбцов

Без `sets` выбираются бинарные числовые/булевы столбцы без пропусков. При наличии бинарных метаданных передавайте `sets` явно.

```python
auto_fig, auto_axes = pyupset.plot(df.assign(label="sample"))
auto_fig
```

## Пустые данные и ошибки

```python
empty, _ = pyupset.plot(df.iloc[:0], sets=list(df.columns))
empty
```

```python
try:
    pyupset.plot(pd.DataFrame({"A": [1, None]}), sets=["A"])
except ValueError as error:
    print(error)
```
